```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LogisticRegression

# 3. Create Custom Data
# [Attendance %, Assignment Grade]
X_log = np.array([[90, 85], [40, 50], [80, 70], [30, 40], [95, 90], [20, 30]])
y_log = np.array([1, 0, 1, 0, 1, 0]) # 1 = Pass, 0 = Fail

model_log = LogisticRegression()
model_log.fit(X_log, y_log)

# Test prediction
test_student = np.array([[75, 65]])
print(f"Prediction for [75% Attendance, 65 Grade]: {'Pass' if model_log.predict(test_student)[0] == 1 else 'Fail'}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 3. Create Custom Data
# X = Study Hours, y = Exam Score
X_simple = np.array([2, 3, 5, 7, 9, 11, 14, 15]).reshape(-1, 1)
y_simple = np.array([50, 55, 62, 70, 78, 85, 92, 95])

# Modeling
model_simple = LinearRegression()
model_simple.fit(X_simple, y_simple)
prediction = model_simple.predict(X_simple)

# Visualization
plt.scatter(X_simple, y_simple, color='blue')
plt.plot(X_simple, prediction, color='red')
plt.title('Study Hours vs Exam Score')
plt.xlabel('Hours')
plt.ylabel('Score')
plt.show()

print(f"R-squared Score: {model_simple.score(X_simple, y_simple):.4f}")
```

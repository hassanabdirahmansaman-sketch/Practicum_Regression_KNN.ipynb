```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# 3. Create Custom Data
# Columns: [Square Footage, Age of House]
X_multi = np.array([[1500, 5], [2000, 10], [1200, 2], [2500, 15], [1800, 7]])
# Price in thousands
y_multi = np.array([300, 380, 270, 450, 340])

model_multi = LinearRegression()
model_multi.fit(X_multi, y_multi)

# Analysis of coefficients
print(f"Coefficients: {model_multi.coef_}")
print(f"Intercept: {model_multi.intercept_}")
```
